```python
# Question 41: Check if string contains balanced parentheses: "((()))"
print("\nQuestion 41: Check if string contains balanced parentheses: '((()))'")
text="((()))"
count=0

for char in text:
    if char=="(":
        count=count+1
    elif char==")":
        count=count-1

if count==0:
    print("Balanced parentheses:",True)
else:
    print("Balanced parentheses:",False)
```

```python
# Question 42: Convert "hello world" to Morse code
print("\nQuestion 42: Convert 'hello world' to Morse code")
text="hello world"
# First I am storing Morse codes in a dictionary and then converting each character into its Morse code.
morse={
    "h":"....",
    "e":".",
    "l":".-..",
    "o":"---",
    "w":".--",
    "r":".-.",
    "d":"-.."
    }

result=""
for char in text:
    if char==" ":
        result=result+"/ "
    else:
        result=result+morse[char]+" "

print("Morse code:",result)
```

```python
# Question 43: Find the longest common substring between "programming" and "grammar"
print("\nQuestion 43: Find the longest common substring between 'programming' and 'grammar'")
text1="programming"
text2="grammar"
longest=""

for i in range(len(text1)):
    for j in range(i+1,len(text1)+1):
        part=text1[i:j]
        if part in text2 and len(part)>len(longest):
            longest=part

print("Longest common substring:",longest)
```

```python
# Question 46: Convert "hello world" to Pig Latin
print("\nQuestion 46: Convert 'hello world' to Pig Latin")
text="hello world"
words=text.split()
result=[]
# First I am splitting the string into words and then moving the first letter to the end and adding "ay".
for word in words:
    result.append(word[1:]+word[0]+"ay")

print("Pig Latin:"," ".join(result))
```

```python
# Question 47: Check if string is a valid IPv4 address: "192.168.1.1"
print("\nQuestion 47: Check if string is a valid IPv4 address: '192.168.1.1'")
ip="192.168.1.1"
parts=ip.split(".")

valid=True

if len(parts)!=4:
    valid=False
else:
    for part in parts:
        if not part.isdigit() or int(part)>255:
            valid=False

print("Valid IPv4 address:",valid)
```

```python
# Question 49: Convert "hello world" to ROT13 encoding
print("\nQuestion 49: Convert 'hello world' to ROT13 encoding")
text="hello world"
result=""

for char in text:
    if char==" ":
        result=result+" "
    else:
        result=result+chr((ord(char)-ord("a")+13)%26+ord("a"))

print("ROT13:",result)
```

```python
# Question 50: Check if string is a valid credit card number: "4532015112830366"
print("\nQuestion 50: Check if string is a valid credit card number: '4532015112830366'")
number="4532015112830366"
total=0

for i in range(len(number)):
    digit=int(number[i])

    if i%2==0:
        digit=digit*2
        if digit>9:
            digit=digit-9

    total=total+digit

print("Valid credit card number:",total%10==0)
```
