```python
# SOLVED EXAMPLE
# --------------
# Question: Find the key with maximum value in a dictionary
print("SOLVED EXAMPLE:")
print("Find the key with maximum value in a dictionary")
scores = {'Alice': 85, 'Bob': 92, 'Charlie': 78, 'Diana': 95, 'Eve': 88}
max_key = max(scores, key=scores.get)
max_value = scores[max_key]
print(f"Dictionary: {scores}")
print(f"Key with maximum value: {max_key}")
print(f"Maximum value: {max_value}")
print("-" * 50)
```

```python
print("\nQuestion 9: Count frequency of each character in string 'hello'")
# First I am checking each character and counting how many times it appears.
text="hello"
frequency={}
for char in text:
    if char in frequency:
        frequency[char]=frequency[char]+1
    else:
        frequency[char]=1
print("Frequency:",frequency)
```

```python
print("\nQuestion 34: Find magnitude of each complex number")
# First I am finding the magnitude using real and imaginary parts.
for key in data:
    number=data[key]
    # Magnitude = square root of (real part squared + imaginary part squared)
    magnitude=(number.real**2+number.imag**2)**0.5
    print(key,"magnitude:",magnitude)
```

```python
print("\nQuestion 44: Find sum of each nested list")
# First I am adding the numbers in each nested list using loops.
data={"matrix":[[1,2],[3,4]],"vector":[5,6,7]}
for row in data["matrix"]:
    total=0
    for number in row:
        total+=number
    print("Matrix row sum:",total)
total=0
for number in data["vector"]:
    total+=number
print("Vector sum:",total)
```
