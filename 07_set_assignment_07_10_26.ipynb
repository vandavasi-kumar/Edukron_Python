# SET DATATYPE ASSIGNMENT
# ======================


```python
print("Question 1: Create a set of first 10 natural numbers")
# First I am using a range to create the first 10 natural numbers.
numbers=set(range(1,11))
print("Set:",numbers)
```

```python
print("\nQuestion 2: Add element 11 to set {1, 2, 3, 4, 5}")
# First I am using add() to add 11 to the set.
numbers={1,2,3,4,5}
numbers.add(11)
print("Set:",numbers)
```

```python
print("\nQuestion 3: Remove element 3 from set {1, 2, 3, 4, 5, 6}")
# First I am using remove() to remove 3 from the set.
numbers={1,2,3,4,5,6}
numbers.remove(3)
print("Set:",numbers)
```

```python
print("\nQuestion 4: Find the intersection of {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8}")
# First I am finding the elements that are common in both sets.
set1={1,2,3,4,5}
set2={4,5,6,7,8}
result=set1.intersection(set2)
print("Intersection:",result)
```

```python
print("\nQuestion 5: Find the difference between {1, 2, 3, 4, 5} and {4, 5, 6, 7, 8}")
# First I am finding the elements that are only present in the first set.
set1={1,2,3,4,5}
set2={4,5,6,7,8}
result=set1.difference(set2)
print("Difference:",result)
```

```python
print("\nQuestion 6: Check if 5 is in set {1, 2, 3, 4, 6, 7, 8}")
# First I am checking whether 5 is present in the set.
numbers={1,2,3,4,6,7,8}
print("5 is present:",5 in numbers)
```

```python
print("\nQuestion 7: Find the length of set {'a', 'b', 'c', 'd', 'e'}")
# First I am using len() to find the number of elements in the set.
letters={"a","b","c","d","e"}
print("Length:",len(letters))
```

```python
print("\nQuestion 8: Create a set of vowels from string 'hello world'")
# First I am checking each character and adding the vowels to the set.
text="hello world"
vowels=set()
for char in text:
    if char in "aeiou":
        vowels.add(char)
print("Vowels:",vowels)
```

```python
print("\nQuestion 9: Remove duplicates from list [1, 2, 2, 3, 4, 4, 5, 6, 6, 7] using set")
# First I am converting the list to a set to remove duplicate values.
numbers=[1,2,2,3,4,4,5,6,6,7]
result=set(numbers)
print("Without duplicates:",result)
```

```python
print("\nQuestion 10: Check if {1, 2, 3} is a subset of {1, 2, 3, 4, 5, 6}")
# First I am checking whether all elements of the first set are present in the second set.
set1={1,2,3}
set2={1,2,3,4,5,6}
print("Is subset:",set1.issubset(set2))
```
