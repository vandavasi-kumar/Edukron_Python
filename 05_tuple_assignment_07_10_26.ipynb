```python
# SOLVED EXAMPLE
# --------------
# Question: Find the sum and product of all elements in a tuple
print("SOLVED EXAMPLE:")
print("Find the sum and product of all elements in a tuple")
numbers = (2, 4, 6, 8, 10)
sum_tuple = sum(numbers)
product = 1
for num in numbers:
    product *= num
print(f"Tuple: {numbers}")
print(f"Sum: {sum_tuple}")
print(f"Product: {product}")
print("-" * 50)
```

```python
print("\nQuestion 3: Access the 3rd element from tuple ('a', 'b', 'c', 'd', 'e')")
# First I am using index 2 because tuple indexing starts from 0.
letters=("a","b","c","d","e")
print("3rd element:",letters[2])
```

```python
print("\nQuestion 4: Find the maximum value in tuple (23, 45, 12, 67, 34, 89, 56)")
# First I am comparing the values to find the largest number.
numbers=(23,45,12,67,34,89,56)
maximum=numbers[0]
for number in numbers:
    if number>maximum:
        maximum=number
print("Maximum value:",maximum)
```

```python
print("\nQuestion 5: Count how many times 5 appears in (1, 5, 2, 5, 3, 5, 4, 5, 6)")
# First I am using the count() method to find how many times 5 appears.
numbers=(1,5,2,5,3,5,4,5,6)
# print("Count of 5:",numbers.count(5))
c=0
for i in numbers:
    if i==5:
        c+=1
print("Count of 5:",c)
        
```

```python
print("\nQuestion 9: Create a tuple of first 5 even numbers")
# First I am creating a tuple with the first five even numbers.
# numbers=(2,4,6,8,10)

numbers=[]

for i in range(1,6):
    numbers.append(i*2)
numbers=tuple(numbers)

print("Tuple:",numbers)
```

```python
print("\nQuestion 10: Find the average of numbers in tuple (15, 23, 31, 42, 56, 78)")
# First I am adding all the numbers and dividing the sum by the number of values.
numbers=(15,23,31,42,56,78)
total=0
for number in numbers:
    total=total+number
average=total/len(numbers)
print("Average:",average)
```
