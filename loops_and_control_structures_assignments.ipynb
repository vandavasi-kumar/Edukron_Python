Algorithm:

4. Repeat until the remainder is 0

### Problem 7.1: Collatz Conjecture (Medium)

The Collatz conjecture is one of the most famous unsolved problems in mathematics. The rules are:
1. Start with any positive integer n
2. If n is even, divide it by 2
3. If n is odd, multiply it by 3 and add 1
4. Repeat until you reach 1

Write a program that:

- Counts how many steps it took

Example: Starting with 6: 6 → 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1 (8 steps)

````python
n = int(input("Enter a positive number: "))
steps = 0
print("Sequence:")
print(n, end="")

while n != 1:
    if n % 2 == 0:
        n = n // 2       
    else:
        n = 3 * n + 1     
        
    print(" ->", n, end="")
    steps += 1

print()  
print(f"Total steps: {steps}")
````

### Problem 7.2: Caesar Cipher (Hard)

Write a program that implements a Caesar cipher to encrypt and decrypt messages.

The Caesar cipher shifts each letter by a fixed number of positions in the alphabet.

Example: With shift=3
- 'A' becomes 'D'
- 'Hello' becomes 'Khoor'

Your program should:
1. Ask for a message and shift value
2. Provide options to encrypt or decrypt
3. Handle both uppercase and lowercase letters
4. Keep non-alphabetic characters unchanged

**Hint:** Loop through each character. Use ord() to get ASCII value, shift it, use chr() to convert back. Use modulo 26 for wrapping. Check if char.isalpha().

````python
message = input("Enter the message: ")
shift = int(input("Enter the shift value: "))

choice = input("Enter E to Encrypt or D to Decrypt: ")

result = ""

if choice.upper() == "D":
    shift = -shift

for char in message:
    if char.isalpha():
        if char.isupper():
            start = ord('A')
        else:
            start = ord('a')

        new_char = chr((ord(char) - start + shift) % 26 + start)
        result += new_char
    else:
        result += char

if choice.upper() == "E":
    print("Encrypted message:", result)
elif choice.upper() == "D":
    print("Decrypted message:", result)
else:
    print("Invalid choice!")
````

**Hint:** Use variables for balance, pin. Loop for PIN attempts. Use while loop for menu. Use list to store transaction history. Validate amounts > 0 and balance >= withdrawal.

````python
balance = 10000
pin = "1234"
attempts = 3
history = []

# PIN verification
while attempts > 0:
    entered_pin = input("Enter your PIN: ")

    if entered_pin == pin:
        print("\nLogin successful!")
        break
    else:
        attempts -= 1
        print("Incorrect PIN. Attempts left:", attempts)

if attempts == 0:
    print("Your account is locked. Please try again later.")

else:
    while True:
        print("\n===== ATM MENU =====")
        print("1. Check Balance")
        print("2. Withdraw Money")
        print("3. Deposit Money")
        print("4. Change PIN")
        print("5. Transaction History")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            print("Current balance:", balance)

        elif choice == "2":
            try:
                amount = float(input("Enter withdrawal amount: "))

                if amount <= 0:
                    print("Amount must be greater than zero.")
                elif amount > balance:
                    print("Insufficient balance.")
                else:
                    balance -= amount
                    print("Withdrawal successful.")
                    print("Remaining balance:", balance)
                    history.append("Withdrawn: " + str(amount))

            except ValueError:
                print("Invalid amount. Enter a number.")

        elif choice == "3":
            try:
                amount = float(input("Enter deposit amount: "))

                if amount <= 0:
                    print("Amount must be greater than zero.")
                else:
                    balance += amount
                    print("Deposit successful.")
                    print("Current balance:", balance)
                    history.append("Deposited: " + str(amount))

            except ValueError:
                print("Invalid amount. Enter a number.")

        elif choice == "4":
            old_pin = input("Enter your current PIN: ")

            if old_pin == pin:
                new_pin = input("Enter your new 4-digit PIN: ")

                if len(new_pin) == 4 and new_pin.isdigit():
                    if new_pin != pin:
                        pin = new_pin
                        print("PIN changed successfully.")
                        history.append("PIN changed")
                    else:
                        print("New PIN must differ from old PIN.")
                else:
                    print("PIN must contain exactly 4 digits.")
            else:
                print("Incorrect current PIN.")

        elif choice == "5":
            print("\n===== LAST 5 TRANSACTIONS =====")

            if len(history) == 0:
                print("No transactions yet.")
            else:
                for transaction in history[-5:]:
                    print(transaction)

        elif choice == "6":
            print("Thank you for using the ATM!")
            break

        else:
            print("Invalid choice. Please select 1 to 6.")

        # Keep only the last 5 transactions
        history = history[-5:]
````

### Problem 8.9: Temperature Converter (Medium)

- F to C: C = (F - 32) × 5/9

````python
choice = int(input("Enter your choice (1 or 2): "))
if choice == 1:
    celsius = float(input("Enter temperature in Celsius: "))
    fahrenheit = (celsius * 9/5) + 32
    print("Temperature in Fahrenheit:", fahrenheit)
elif choice == 2:
    fahrenheit = float(input("Enter temperature in Fahrenheit: "))
    celsius = (fahrenheit - 32) * 5/9
    print("Temperature in Celsius:", celsius)
else:
    print("Invalid choice!")
````
